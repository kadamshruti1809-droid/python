```python

print("====== WELCOME TO OUR HOTEL ======")
print("1. Pizza      - Rs. 250")
print("2. Burger     - Rs. 120")
print("3. Sandwich   - Rs. 100")
print("4. Coffee     - Rs.  80")

choice = int(input("Enter your food choice: "))
quantity = int(input("Enter quantity: "))

if choice == 1:
    price = 250
elif choice == 2:
    price = 120
elif choice == 3:
    price = 100
elif choice == 4:
    price = 80
else:
    price = 0
    print("Invalid food choice!")

if price > 0 and quantity > 0:
    total = price * quantity

    if total >= 1000:
        discount = total * 0.10
    else:
        discount = 0

    final_bill = total - discount

    print("\n====== HOTEL BILL ======")
    print("Price per item: Rs.", price)
    print("Quantity:", quantity)
    print("Total amount: Rs.", total)
    print("Discount: Rs.", discount)
    print("Final bill: Rs.", final_bill)
    print("========================")
```

```python
def cube(c):
    x= c**3
    return x
l1=[2,4,7,9]
result= list(map(cube,l1)) # map,function,iterable
print(result)
```

```python
# WAP to print even or odd.
# E.G: l1=[2,4,7,9]
# answer :["even",even odd,odd]


def check(x):
    if x%2==0 :
        return("even")
    else:
        return("odd")    
l1=[2,4,7,9]
result=list(map(check,l1))
print(result)
```

# Inheritance

```python
class employee ():
    def __init__(self,ID, name ,addhar):
        print("parent constuctor has run")
        self.Id = ID
        self.Name = name
        self.Adhar = addhar
    def Welcome(self):
        print(f"welcome {self.Name}")
#-----------------------------------
class department(employee): # CHILD
    def __init__(self, ID,name, addhar, dept,salary):
        print("child consturctor has run")
        #super function calls the parent's constructor
        super().__init__ (ID ,name ,addhar) # will read all these attributes
        # From the parent class 

        self.depatment = dept
        self. Payment = salary 
    def info(self):
        print(f"the department of {self.Name} is {self.depatment}")
        print(f"The diwali  bonus of {self.Name} is {(self.Payment*0.05)} ")
```

```python
p1 = employee(10002 , "John Doe", 14561237)
```

```python
c1 = department(1000,"Mary Smith ", 45698741 , "HR", 45000)
```

```python
p1.Welcome()
c1.info()
```

```python
p1.info() #parent class dosen't have access to child clSSS
```

```python
c1.Welcome() # child class have access of parent class but parent dosen't have.
```

```python
class employee ():
    def __init__(self,ID, name ,addhar):
        print("parent constuctor has run")
        self.Id = ID
        self.Name = name
        self.Adhar = addhar
    def info(self):
        print(f"welcome {self.Name}")
#-----------------------------------
class department(employee): # CHILD
    def __init__(self, ID,name, addhar, dept,salary):
        print("child consturctor has run")
        #super function calls the parent's constructor
        super().__init__ (ID ,name ,addhar) # will read all these attributes
        # From the parent class 

        self.depatment = dept
        self. Payment = salary 
    def info(self):      #method overrinding ocurrse when child classs defines a method inherites parent class .
        print(f"welcome to the company {self.Name}")
        print(f"the department of {self.Name} is {self.depatment}")
        print(f"The diwali  bonus of {self.Name} is {(self.Payment*0.05)} ")
p1 = employee(10002 , "John Doe", 14561237)
c1 = department(1000,"Mary Smith ", 45698741 , "HR", 45000)
p1.info()
c1.info()
```

# Abstraction


```python
from abc import ABc,abstractmethod
class company (ABC): # Abstract -Parent class
    # -----Abstract Method ------
    @abstarctmethods
    def info (self): #Abstract method
        print("Information")
    def details(self): # MEthod 2
        print("details")
```

```python
from abc import ABC, abstractmethod


class company(ABC): # Abstract - Parent class
# --------------Abstract Method---------------
    @abstractmethod
    def info(self): # Abstract Method 1
        print("Information")
    def details(self): # Method 2
        print("Details")
#--------------------------
class dept_HR(company): # child class
    def KPI(self):
        print("KPI")
    def info(self):
        pass
    
#-------------------------
IT_1001=dept_HR() # object creation
IT_1001.KPI() # object cannot use the class unless "info" method is implemented in child class
```
